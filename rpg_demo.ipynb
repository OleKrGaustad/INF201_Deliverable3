### 2.1 Dice class

**Purpose:** Introduce randomness into damage calculations.

**Attributes:**
- `sides` (int) – number of sides on the dice (≥ 2)

**Methods:**
- `roll()` → returns a random int between 1 and `sides`
- `__call__()` → allows `d20()` instead of `d20.roll()`
- `__repr__()` → nice string representation

```python
print("=== Dice class overview ===")
print("Public members:")
for name, member in inspect.getmembers(Dice):
    if name in ("__init__", "__repr__", "__call__", "roll") or (not name.startswith("_")):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example ---")
d6 = Dice()              # default 6-sided
d20 = Dice(sides=20)
d4 = Dice(4)

print(d6)
print(d20)
print(d4)
print()
print("d6.roll() →", d6.roll())
print("d20()     →", d20())      # uses __call__
print("d4()      →", d4())
print()
print("Attributes of d20:", vars(d20))
```

### 2.2 Weapon class

**Purpose:** Represent a weapon that deals variable damage using a Dice.

**Attributes:**
- `name` (str)
- `damage_dice` (Dice)

**Methods:**
- `roll_damage()` → rolls the weapon’s dice and returns the damage
- `__repr__()` / `__str__()`

```python
print("=== Weapon class overview ===")
print("Public members:")
for name, member in inspect.getmembers(Weapon):
    if name in ("__init__", "__repr__", "__str__", "roll_damage") or (not name.startswith("_")):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example ---")
longsword = Weapon("Longsword", Dice(8))
dagger = Weapon("Dagger", Dice(4))
fire_staff = Weapon("Fire Staff", Dice(10))

print(longsword)
print(dagger)
print(fire_staff)
print()
print("Damage rolls:")
print("  Longsword :", longsword.roll_damage())
print("  Dagger    :", dagger.roll_damage())
print("  Fire Staff:", fire_staff.roll_damage())
print()
print("Attributes of longsword:", vars(longsword))
```

### 2.3 Character class (base class)

**Purpose:** Base class for all characters. Handles HP, basic attacks and encapsulation.

**Attributes:**
- `name` (str)
- `_hp` (private) – current hit points
- `hp_max` (int)
- `damage` (int) – base damage used when no weapon is equipped
- `weapon` (Weapon | None)

**Properties (read-only):**
- `hp` → current hit points
- `is_alive` → `True` if `_hp > 0`

**Methods:**
- `take_damage(amount)` – reduces own HP (never below 0)
- `heal(amount)` – restores HP (never above `hp_max`)
- `attack(target)` – deals damage to another character (uses weapon if equipped)
- `introduce()`
- `__repr__()` / `__str__()`

```python
print("=== Character class overview ===")
print("Public members:")
for name, member in inspect.getmembers(Character):
    if name in ("__init__", "__repr__", "__str__", "attack", "heal",
                "take_damage", "introduce", "hp", "is_alive") or (not name.startswith("_")):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example ---")
hero = Character("Aria", hp_max=100, damage=12, weapon=longsword)
goblin = Character("Goblin", hp_max=40, damage=6)   # no weapon

print(hero)
print(goblin)
print()
print("hero.hp        →", hero.hp)
print("hero.is_alive  →", hero.is_alive)
print("hero.weapon    →", hero.weapon)
print()
print("--- Hero attacks Goblin ---")
hero.attack(goblin)
print(goblin)
print()
print("--- Goblin takes heavy damage ---")
goblin.take_damage(50)
print(goblin)
print("goblin.is_alive →", goblin.is_alive)
```

### 2.4 Warrior class (subclass of Character)

**Purpose:** Defensive specialist.

**New attribute:**
- `armor` (int, default = 5)

**Overridden method:**
- `take_damage(amount)` – incoming damage is reduced by `armor` before being applied

Inherits everything else from `Character` (`hp`, `is_alive`, `attack`, `heal`, etc.).

```python
print("=== Warrior class overview ===")
print("Public members:")
for name, member in inspect.getmembers(Warrior):
    if name in ("__init__", "__repr__", "__str__", "attack", "heal",
                "take_damage", "introduce", "hp", "is_alive") or (not name.startswith("_")):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example ---")
warrior = Warrior("Thorgar", hp_max=120, damage=14, weapon=longsword, armor=7)
print(warrior)
print()
print("warrior.armor →", warrior.armor)
print()
print("Warrior takes 20 raw damage (armor reduces it by 7):")
print("HP before:", warrior.hp)
warrior.take_damage(20)
print("HP after :", warrior.hp)   # 120 - (20-7) = 107
```

### 2.5 Mage class (subclass of Character)

**Purpose:** Resource-based magical attacker.

**New attributes:**
- `mana` (int)
- `mana_max` (int, default = 30)

**Overridden method:**
- `attack(target)`:
  - If `mana >= 10` → spend 10 mana and deal **double damage** (magic attack, ignores weapon)
  - If `mana < 10` → normal attack (uses weapon if equipped) **and restore 5 mana**

```python
print("=== Mage class overview ===")
print("Public members:")
for name, member in inspect.getmembers(Mage):
    if name in ("__init__", "__repr__", "__str__", "attack", "heal",
                "take_damage", "introduce", "hp", "is_alive") or (not name.startswith("_")):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example ---")
mage = Mage("Elara", hp_max=80, damage=10, weapon=fire_staff, mana_max=30)
print(mage)
print()
print("mage.mana     →", mage.mana)
print("mage.mana_max →", mage.mana_max)
print()

dummy = Character("Training Dummy", hp_max=200, damage=1)

print("--- Mage attacks (shows mana system) ---")
for i in range(4):
    print(f"Attack {i+1}: mana before = {mage.mana}")
    mage.attack(dummy)
    print(f"           mana after  = {mage.mana}, Dummy HP = {dummy.hp}")
    print()
```

### 2.6 Battle class

**Purpose:** Manage turn-based combat between two characters.

**Attributes:**
- `hero` (Character)
- `enemy` (Character)
- `round_number` (int)

**Methods:**
- `play_round()` – hero attacks, then enemy attacks (if still alive)
- `simulate_battle(verbose=True)` – runs rounds until one character is defeated and returns the winner

```python
print("=== Battle class overview ===")
print("Public members:")
for name, member in inspect.getmembers(Battle):
    if name in ("__init__", "__repr__", "play_round", "simulate_battle") or (not name.startswith("_")):
        print(f"  • {name}: {type(member).__name__}")

print("\n--- Live example (short fight) ---")
fighter = Character("Rogan", hp_max=60, damage=15, weapon=Weapon("Axe", Dice(8)))
bandit = Character("Bandit", hp_max=45, damage=8)

battle = Battle(hero=fighter, enemy=bandit)
print(battle)
print()
winner = battle.simulate_battle(verbose=True)
print(f"\nWinner: {winner.name}")
```

## 3. Simulated Battles

### 3.1 Warrior vs Mage (both with weapons)

```python
hero = Warrior(
    "Thorgar",
    hp_max=130,
    damage=13,
    weapon=Weapon("Longsword", Dice(8)),
    armor=6
)

enemy = Mage(
    "Dark Elara",
    hp_max=95,
    damage=11,
    weapon=Weapon("Arcane Staff", Dice(6)),
    mana_max=40
)

battle1 = Battle(hero=hero, enemy=enemy)
winner1 = battle1.simulate_battle(verbose=True)

print(f"\nWinner of Battle 1: {winner1.name}")
```

### 3.2 Two base Characters (one without weapon)

```python
fighter = Character("Rogan", hp_max=100, damage=15, weapon=Weapon("Battle Axe", Dice(10)))
bandit = Character("Bandit", hp_max=70, damage=9)  # no weapon

battle2 = Battle(hero=fighter, enemy=bandit)
winner2 = battle2.simulate_battle(verbose=True)

print(f"\nWinner of Battle 2: {winner2.name}")
```

### 3.3 Mage with low starting mana (demonstrates fallback attack)

```python
mage2 = Mage("Lyra", hp_max=85, damage=12, weapon=Weapon("Crystal Wand", Dice(4)), mana_max=25)
mage2.mana = 5   # force low mana so we see the fallback + mana restore

warrior2 = Warrior("Borin", hp_max=110, damage=11, weapon=Weapon("Warhammer", Dice(8)), armor=5)

print("Starting state:")
print(f"  {mage2}")
print(f"  {warrior2}")
print()

battle3 = Battle(hero=mage2, enemy=warrior2)
winner3 = battle3.simulate_battle(verbose=True)

print(f"\nWinner of Battle 3: {winner3.name}")
```
