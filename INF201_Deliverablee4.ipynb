## 1. Temperature

The `Temperature` class is imported from `Temperature.py`.
The examples below check the implemented methods manually. The complete automated tests are in `Test_temperature.py`.

```python
from Temperature import Temperature
```

```python
# Createss temperatures in different units
celsius = Temperature(25, "C")
fahrenheit = Temperature(77, "F")
kelvin = Temperature(298.15, "K")

print(celsius)
print(fahrenheit)
print(kelvin)
```

```python
# Convert between units
print(celsius.to_fahrenheit())
print(celsius.to_kelvin())
print(celsius.convert_to("F"))
```

```python
# Compare temperatures expressed in different units
print(Temperature(0, "C") == Temperature(32, "F"))
print(Temperature(0, "C") < Temperature(50, "F"))
```

### 1.1 tests with pytest

The tests are written in the separate file `Test_temperature.py`. That file imports `pytest` and `Temperature`.

```python
!pytest -v Test_temperature.py
```

## 2. Fraction

The `Fraction` class is imported from `Fraction.py`.
The examples below check initialization, fraction simplification, comparisons, and the four spesial methods implemented in the class.

```python
from Fraction import Fraction
```

```python
# Create and maek fractions simpler
first = Fraction(6, 8)
second = Fraction(1, 3)

print(first)
print(second)
print(repr(first))
```

```python
# Compare fractions
print(Fraction(1, 2) == Fraction(2, 4))
print(Fraction(1, 3) < Fraction(1, 2))
print(Fraction(3, 4) > Fraction(1, 2))
```

```python
# spesials
a = Fraction(1, 2)
b = Fraction(1, 3)

print(a + b)
print(a - b)
print(Fraction(2, 3) * Fraction(3, 4))
print(Fraction(1, 2) / Fraction(1, 4))
```

### 2.1 tests with pytest

The tests are written in the separate file `Test_Fraction.py`. That file imports `pytest` and `Fraction`, and contains the test functions. This command runs the test script.

```python
!pytest -v Test_Fraction.py
```
